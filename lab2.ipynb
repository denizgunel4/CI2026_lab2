```python
from random import random, sample, seed
from icecream import ic
```

```python
N = 10

seed(42)

MAX_FAILED_ATTEMPTS = 25
```

```python
OBJECTS = {n for n in range(N)}


def create_instance():
    sets = tuple(frozenset(sample(range(N), k=s+1)) for s in range(N))
    costs = tuple(10*random()+(s+random())**2 for s in range(N))

    return sets, costs

SETS, COSTS = create_instance()
```

```python
def fitness(solution):
    cost = sum(COSTS[i] for i in range(N) if solution[i])
    return -cost

def is_cover(solution):
    selected_sets = [SETS[i] for i in range(N) if solution[i]]
    covered = frozenset().union(*selected_sets)

    return covered == OBJECTS
```

```python
def tweak(solution):
    indices = sample(range(N), k=2)

    candidate = list(solution)

    for index in indices:
        candidate[index] = not candidate[index]

    return tuple(candidate)
```

```python
def solve():
    failed_attempts = 0
    total_steps = 0
    
    solution = tuple(sample([True, False], k=N, counts=(N, N)))

    while not is_cover(solution):
        solution = tuple(sample([True, False], k=N, counts=(N, N)))

    while failed_attempts < MAX_FAILED_ATTEMPTS:
        total_steps += 1

        candidate = tweak(solution)

        step = total_steps

        if is_cover(candidate) and fitness(candidate) > fitness(solution):
            solution = candidate
            failed_attempts = 0
            status = "ACCEPTED"
            ic(step, failed_attempts, candidate, fitness(candidate), is_cover(candidate), status)
        else:
            failed_attempts += 1
            status = "REJECTED"
            ic(step, failed_attempts, candidate, fitness(candidate), is_cover(candidate), status)

    return solution, -fitness(solution), total_steps
```

```python
results = []

for instance in range(100):
    solution, cost, steps = solve()
    results.append((solution, cost, steps))
```

```python
for instance, (solution, cost, steps) in enumerate(results, start=1):
    print(f"Instance {instance}")
    print("Solution:", solution)
    print("Cost:", cost)
    print("Steps:", steps)
    print()
```
